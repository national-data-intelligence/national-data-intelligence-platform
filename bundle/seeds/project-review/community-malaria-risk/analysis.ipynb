# Community Malaria Outbreak-Risk — exploratory analysis

Reviewer notebook: pulls the feature mart, inspects the weekly malaria
signal per woreda, and sanity-checks the Farrington baseline before the
model is graduated.

```python
import os, yaml, pandas as pd, sqlalchemy
spec = yaml.safe_load(open('model/feature_spec.yaml'))
eng = sqlalchemy.create_engine(os.environ['OPS_DB_URL'])
df = pd.read_sql(spec['input']['query'], eng)
df.shape
```

```python
# weekly malaria cases across woredas — the signal the detector watches
wk = df.groupby(['year','week'])['cases'].sum().reset_index()
wk.tail(12)
```

# Community Stockout Forecast — exploratory analysis

Reviewer notebook: pulls the monthly supply mart and inspects
consumption (AMC) vs stock-on-hand trends per commodity.

```python
import os, yaml, pandas as pd, sqlalchemy
spec = yaml.safe_load(open('model/feature_spec.yaml'))
eng = sqlalchemy.create_engine(os.environ['OPS_DB_URL'])
df = pd.read_sql(spec['input']['query'], eng)
df.groupby('commodity_name')['amc'].mean().sort_values(ascending=False).head(10)
```
